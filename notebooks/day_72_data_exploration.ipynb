# Day 72: Data Exploration with Pandas

Do college degrees pay you back? We explore the PayScale post-university salary data by undergraduate major and answer five questions:

1. Which degrees have the highest starting salaries?
2. Which majors have the lowest earnings after college?
3. Which degrees have the highest earning potential?
4. What are the lowest-risk majors from an earnings standpoint?
5. Do Business, STEM or HASS degrees earn more on average?

Every step calls a tested function from `college_majors.application.lecture_steps`.

```python
import pandas as pd

from college_majors.application import lecture_steps as steps
from college_majors.infrastructure.csv_repository import read_data

pd.options.display.float_format = '{:,.2f}'.format
pd.set_option('display.width', 200)
pd.set_option('display.max_columns', None)
```

## Step 1: Read the .csv file and look at the first rows

`pd.read_csv` loads the file; `df.head()` shows the first 5 rows.

```python
df = read_data()
steps.head(df)
```

## Step 2: Shape and column names

`df.shape` gives rows and columns. `df.columns` gives the column names.

```python
print(steps.shape(df))
steps.columns(df)
```

## Step 3: Missing values

`df.isna()` marks every cell that is NaN. Looking at `df.tail()` shows a source row with blank values.

```python
steps.missing_values(df).tail()
```

```python
steps.tail(df)
```

## Step 4: Clean the data

`df.dropna()` removes the junk last row and leaves 50 majors.

```python
clean_df = steps.clean(df)
print(steps.shape(clean_df))
steps.tail(clean_df)
```

## Step 5: Highest starting salary

`max()` finds the value, `idxmax()` finds the row index and `.loc` looks up the major.

```python
steps.highest_starting_salary(clean_df)
```

## Step 11: Answers to the five questions

The function `answer_five_questions` collects everything we found. Each question gets a table and one clear answer.

```python
from college_majors.application.answers import answer_five_questions

answers = answer_five_questions(clean_df)
```

### 1. Which degrees have the highest starting salaries?

```python
answers.highest_starting
```

**Answer:** Physician Assistant by far (74,300), followed by Chemical, Computer, Electrical and Mechanical Engineering (57,900 to 63,200).

### 2. Which majors have the lowest earnings after college?

```python
print('Lowest starting salaries')
print(answers.lowest_starting)
print()
print('Lowest mid-career salaries')
print(answers.lowest_mid_career)
```

**Answer:** Spanish (34,000), Religion and Education start lowest. In mid-career Education and Religion earn least (52,000), followed by Spanish, Interior Design and Music.

### 3. Which degrees have the highest earning potential?

```python
answers.highest_potential
```

**Answer:** Economics (210,000 at the 90th percentile), Finance, Chemical Engineering, Math and Physics.

### 4. What are the lowest-risk majors from an earnings standpoint?

```python
answers.lowest_risk
```

**Answer:** Nursing has the smallest spread between low and high earners (50,700), then Physician Assistant, Nutrition, Spanish and Health Care Administration.

### 5. Do Business, STEM or HASS degrees earn more on average?

```python
print(answers.group_averages)
print()
print(f'Best-earning group: {answers.best_group}')
```

**Answer:** STEM earns most on average (about 53,900 at the start and 90,800 in mid-career), then Business (44,600 and 75,100), then HASS (37,200 and 63,000).
